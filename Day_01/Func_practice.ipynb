```python
def sum(a,b):
    return (a+b)

sum(3,5)
```

```python
def max(a,b):
    if (a<b):
        return f"{b} is greater"

    if (a>b):
        return f"{a} is greater"

max(3,5)
```

```python
def list(a,b,c):
    return (a+b+c)

list(1,2,3)
```

```python
def first_function(a):
    return (a*a)

def second_function(b,c):
    return (first_function(b) + first_function(c))

print(second_function(3,4))
```

```python
def greeting(name):
    return f"hello {name}"

def second_greeting(name):
    return greeting(name) + " how are you"

print(second_greeting("Gokul"))
```
