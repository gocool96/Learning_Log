```python
class car:
                                                                                                                                                                                                                                                                                                                                                                                         wheels = 4
    def __init__(self,brand,make,speed = 0):
        self.brand = brand
        self.make = make
        self.speed = speed

    def acclerate(self,amount):
        self.speed+=amount
        print(f"{self.brand} {self.make} is going {self.speed} km/h")

    def brake(self,amount):
        self.speed -=amount
        if self.speed < 0:
            self.speed = 0
        print(f"{self.brand} {self.make} is slowed to {self.speed} km/h")

    def honk(self):
        print(f"{self.brand} {self.make} says beep beep")

    def fullstop(self):
        self.speed = 0
        print(f"{self.brand} {self.make} has stopped.")
car1 = car("Ford","Mustang",speed = 100)
car2 = car("BMW","M3",speed = 100)

car1.acclerate(10)
car2.acclerate(20)
car1.honk()
car1.brake(100)
car2.fullstop()
```
