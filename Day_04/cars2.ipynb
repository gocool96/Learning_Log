```python
class car:

    wheels = 4

    
    def __init__(self,brand,make,speed = 0):
        self.brand = brand
        self.make = make
        self.speed = speed

    def __str__(self):
        return f"{self.brand} {self.make}, currently at {self.speed} km/h"

    def acclerate(self,amount):
        self.speed  += amount
        print(f"{self.brand}{self.make} has acclerated to {self.speed} km/h")

    def brake(self,amount):
        self.speed-=amount
        if self.speed < 0:
            self.speed = 0
        print(f"{self.brand}{self.make} has slowed down to {self.speed} km/h")

    def honk(self):
        print(f"{self.brand}{self.make} says beep beep")

    def fullstop(self):
        self.speed = 0
        print(f"{self.brand} {self.make} has stopped.")


class ElectricCar(car):
    def __init__(self,brand,make,speed,battery_level):
        super().__init__(brand,make,speed)
        self.battery_level = battery_level

    def charge(self):
        self.battery_level = 100
        print(f"{self.brand}{self.make} is fully charged")

tesla = ElectricCar("Tesla", "Model 3", 0, 50)


car1 = car("Ford"," Mustang",30)
car2 = car("BMW"," M3",40)
tesla.acclerate(20)
tesla.charge()
car1.acclerate(10)
car2.brake(10)
car1.honk()
car2.fullstop()
print(car1)
print(tesla)
```
